## Tensor is a multi dimension array, it is used to storage the data and calculate on CPU/GPU
   - ex: scaler(0D): 3: number
   - ex: vector(1D): [1,2,3] :array (x,y,z)
   - ex: matrix(3D): [[1,2],[3,4]] :tabular
   - ex: Tensor(nD): (N,H,C,M,..) data for ML

```python
x = torch.zeros(5,3, dtype = torch.long) #
x = torch.Tensor([5.5, 3.3]) # only 1 rows, divide by 'commas', ex: ([....], [...])
x = torch.Tensor(5,3) # rubbish value with 5 rows and col
x
```

```python
y = torch.randn_like(x, dtype = torch.double) # make y tensor size like x (_like),
# with value(randn...) and uses dtype = torch.double..
y
```

```python
y.shape
y.size()
```

## Calculation with pytorch

```python
z = x + y #normal 
z = torch.add(x, y) #use method in torch
z
```

```python
y.add_(x) #make individual and assume later
z = y
z
```

```python
z = torch.empty(5,3) # make dataframe
torch.add(x,y, out = z) # use special method
z
```

## Index with pytorch

## Move to cal on gpu

```python
device = torch.cuda.is_available()
device
```

```python
if torch.cuda.is_available (): #if compyter have gpu -> do this
    device = torch.device("cuda") # cuda = gpu -> point the tensor runtime
    y = torch.ones_like(x, device = device) # same randn, make code fill with ones, shpe like x
    # run on cuda
    x = x.to(device) # x from cpu, transfer it to cuda to cal
    # want to plus x and y, it is same place as cuda as cpu
    z = x + y
    print(z)

# to the architecture of computer, the calculation on gpu always better with the parralel cal
```
